```python
arr = [10, 13, 5, 23, 3]

largest = arr[0]

for num in arr:
    if largest < num:
        largest = num
print(largest)
```

```python
smallest = arr[0]

for num in arr:
    if num < smallest:
        smallest = num
print(smallest)
```
